```python
class Polygons:
  def __init__(self, name, sides):
    self.name = name
    self.sides = sides

  def display(self):
    print(f"name: {self.name}, sides: {self.sides}")
```

```python
triangle = Polygons("triangle", 3)
triangle.display()
```

```python
#Traditional method overloading
class Claculator:

  def add(self,a,b):
    return a+b

  def add(self,a,b,c):
    return a+b+c
```

```python
obj = Claculator()
obj.add(1,2)
```

```python
class OTTPsubscription:
  def __init__(self, subscription_id,plan, total_payment):
    self.id = subscription_id
    self.plan_package = plan
    self.total_payment = total_payment

  def subscription(self):
    print(f"Subscriber with {self.id} id subscribed to the {self.plan_package}")

  def unsubscribe(self):
    print(f"subscriber with {self.id} unsubscribed to the {self.plan_package}")
```

```python
class PerminumSubscription(OTTPsubscription):
  def __init__(self, subscription_id, plan, total_payment, screens):
    super().__init__(subscription_id, plan, total_payment)
    self.max_screen = screens
  def set_max_screen(self, screens):
    self.max_screen = screens
    print(f"maximum screen set to {self.max_screen} in the perminum plan")
```

```python
obj = PerminumSubscription(1222, "Pro",800000, 7)
```

```python
obj.set_max_screen(9)
```

```python
obj.subscription()
```

```python
obj.unsubscribe()
```

```python
#class task

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def show_person(self):
        print("Name:", self.name)
        print("Age:", self.age)


class Student(Person):
    def __init__(self, name, age, student_id):
        Person.__init__(self, name, age)
        self.student_id = student_id

    def study(self):
        print(f"{self.name} is studying.")
        print("Student ID:", self.student_id)


class Teacher(Person):
    def __init__(self, name, age, subject):
        Person.__init__(self, name, age)
        self.subject = subject

    def teach(self):
        print(f"{self.name} is teaching {self.subject}.")


class TeachingAssistant(Student, Teacher):
    def __init__(self, name, age, student_id, subject, department, salary):
        Student.__init__(self, name, age, student_id)
        Teacher.__init__(self, name, age, subject)
        self.department = department
        self.salary = salary

    def assistant_info(self):
        print("Department:", self.department)
        print("Salary:", self.salary)


# Create object
ta = TeachingAssistant(
    "Relish",
      19,
    "ST101",
    "Python",
    "Computer Science",
    30000
)

ta.show_person()
print()

ta.study()
print()

ta.teach()
print()

ta.assistant_info()
```
