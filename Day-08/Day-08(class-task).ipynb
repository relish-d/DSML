```python
#cheak number is positive and even

num = int(input("Enter a number: "))
if num > 0 and num % 2 == 0:
    print("The number is positive and even.")   
else:
    print("The number is either not positive or not even.")
    
```

```python
#cheak user if logged in or not

is_logged_in = True
if is_logged_in:
    print("User is logged in.")
else:
    print("User is not logged in.")
```

```python
#cheak if age is between 18 and 60 using and

age = int(input("Enter your age: "))
if age >= 18 and age <= 60:
    print("You are between 18 and 60.")
else:
    print("You are not between 18 and 60.")
```

```python
#cheak if a day is saturday or sunday using or

day = input("Enter a day of the week: ")
if day.lower() == "saturday" or day.lower() == "sunday":
    print("It's a weekend!")
else:
    print("It's a weekday.")
```
