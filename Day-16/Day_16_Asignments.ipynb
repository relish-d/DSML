```python
#Q1 Write a program that takes two numbers as input and handles the ZeroDivisionError when dividing the first number by the second.

try:
    num1 = int(input("Enter first number: "))
    num2 = int(input("Enter second number: "))

    result = num1 / num2
    print("Result =", result)

except ZeroDivisionError:
    print("Error: Division by zero is not allowed.")
```

```python
#Q2 Write a program that asks the user for an integer and handles the ValueError if the user enters a non-numeric value.

try:
    num = int(input("Enter an integer: "))
    print("You entered:", num)

except ValueError:
    print("Invalid input! Please enter an integer.")
```

```python
#Q3 Write a program that accesses an element from a list using a user-provided index and handles the IndexError.

numbers = [10, 20, 30, 40, 50]

try:
    index = int(input("Enter index: "))
    print("Element =", numbers[index])

except IndexError:
    print("Error: Index out of range.")
```

```python
#Q4 Write a program that reads a key from a dictionary and handles the KeyError if the key does not exist.

student = {
    "name": "Relish",
    "age": 18,
    "faculty": "BSc IT"
}

try:
    key = input("Enter key: ")
    print(student[key])

except KeyError:
    print("Key does not exist.")
```

```python
#Q5 Write a program that uses try, except, and finally to demonstrate that the finally block always executes.

try:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))

    print("Result =", a / b)

except ZeroDivisionError:
    print("Cannot divide by zero.")

finally:
    print("Finally block always executes.")
```

```python
#Q6 Write a program that attempts to open a file specified by the user and handles the FileNotFoundError

filename = input("Enter file name: ")

try:
    file = open(filename, "r")
    print(file.read())
    file.close()

except FileNotFoundError:
    print("File not found.")
```

```python
#Q7 Write a program that accepts a student's marks. Raise a ValueError if the marks are less than 0 or greater than 100.

try:
    marks = int(input("Enter marks: "))

    if marks < 0 or marks > 100:
        raise ValueError("Marks must be between 0 and 100.")

    print("Marks =", marks)

except ValueError as e:
    print(e)
```

```python
#Q8 Write a program that asks the user to enter their age. Raise a ValueError if the age is negative

try:
    age = int(input("Enter your age: "))

    if age < 0:
        raise ValueError("Age cannot be negative.")

    print("Age =", age)

except ValueError as e:
    print(e)
```

```python
#Q9 Create a custom exception called InvalidPasswordError. Raise it if the password entered by the user is less than 8 characters long.

class InvalidPasswordError(Exception):
    pass

try:
    password = input("Enter password: ")

    if len(password) < 8:
        raise InvalidPasswordError("Password must be at least 8 characters.")

    print("Password accepted.")

except InvalidPasswordError as e:
    print(e)
```

```python
#Q10 Create a custom exception called NegativeNumberError. Raise it when the user enters a negative number and display an appropriate message.

class NegativeNumberError(Exception):
    pass

try:
    num = int(input("Enter a number: "))

    if num < 0:
        raise NegativeNumberError("Negative numbers are not allowed.")

    print("Number =", num)

except NegativeNumberError as e:
    print(e)
```

```python
#Q11 Write a program that repeatedly asks the user for input until a valid integer is entered using exception handling.

while True:
    try:
        num = int(input("Enter an integer: "))
        print("You entered:", num)
        break

    except ValueError:
        print("Invalid input! Try again.")
```

```python
#Q12 Develop a mini ATM program that uses custom exceptions, try-except, and finally to handle deposits, withdrawals, and balance checks safely.

class InsufficientBalanceError(Exception):
    pass

balance = 10000

while True:
    print("\n----- ATM MENU -----")
    print("1. Deposit")
    print("2. Withdraw")
    print("3. Check Balance")
    print("4. Exit")

    choice = input("Enter your choice: ")

    try:
        if choice == "1":
            amount = float(input("Enter deposit amount: "))
            balance += amount
            print("Deposit Successful.")

        elif choice == "2":
            amount = float(input("Enter withdrawal amount: "))

            if amount > balance:
                raise InsufficientBalanceError("Insufficient Balance.")

            balance -= amount
            print("Withdrawal Successful.")

        elif choice == "3":
            print("Current Balance =", balance)

        elif choice == "4":
            print("Thank you for using the ATM.")
            break

        else:
            print("Invalid Choice.")

    except ValueError:
        print("Please enter a valid amount.")

    except InsufficientBalanceError as e:
        print(e)

    finally:
        print("Transaction Completed.")
```
