```python
def factorial(n):
  if n==0 or n==1:
    return 1
  else:
    return n*factorial(n-1)
```

```python
factorial(5)
```

```python
def recur_fibo(n):
  if n<=1:
    return n
  else:
    return(recur_fibo(n-1)+recur_fibo(n-2))
```

```python
nterms = int(input("Enter a number"))

if nterms <=0:
  print("please enter a postive integer")
else:
  print("Fibonacci sequence:")
  for i in range(nterms):
    print(recur_fibo(i))
```

```python
# nested function
def outer(x):
  def inner(y):
    return x+y
  return inner
```

```python
add_five = outer(5)
```

```python
add_five
```

```python
result = add_five(6)
result
```

```python
def myWrapper(func):
  def myInnerFunc():
    print("Inner wrapper.")
    func()
  return myInnerFunc
```

```python
@myWrapper
def myFunc():
  print("Hello, world!")
```

```python
myFunc()
```

```python
def logger(func):
  def wrapper(*args, **kwargs):
    print(f"Calling {func.__name__}")
    return func(*args, **kwargs)
  return wrapper
```

```python
@logger
def greet(name,age):
  print(f"Hello {name}, Age: {age},")
```

```python
greet("Relish",19)
```
