```python
#student result management system

def student_result():
    students = {}

    # Input
    for i in range(5):
        print(f"\nStudent {i+1}")
        name = input("Enter student name: ")
        marks = float(input("Enter marks: "))

        students[name] = marks

    print("\n----- Student Results -----")

    total = 0
    highest = -1
    lowest = 101

    for name, marks in students.items():

        if marks >= 80:
            grade = "A"
        elif marks >= 60:
            grade = "B"
        elif marks >= 40:
            grade = "C"
        else:
            grade = "F"

        print(f"Name: {name}")
        print(f"Marks: {marks}")
        print(f"Grade: {grade}")
        print()

        total += marks

        if marks > highest:
            highest = marks

        if marks < lowest:
            lowest = marks

    average = total / len(students)

    print("Highest Marks =", highest)
    print("Lowest Marks =", lowest)
    print("Average Marks =", average)


student_result()
```

```python
#smart shopping cart system

def shopping_cart():
    prices = []

    # Input prices
    for i in range(5):
        price = float(input(f"Enter price of product {i+1}: "))
        prices.append(price)


    total = 0

    for price in prices:
        total += price


    if total >= 5000:
        discount = total * 0.15
    elif total >= 3000:
        discount = total * 0.10
    elif total >= 1000:
        discount = total * 0.05
    else:
        discount = 0

    final_amount = total - discount


    print("\n----- Shopping Cart -----")
    print("Product List:", prices)
    print("Total Amount:", total)
    print("Discount Amount:", discount)
    print("Final Payable Amount:", final_amount)


shopping_cart()
```
