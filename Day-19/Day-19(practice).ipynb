```python
#Write a program to create a Mobile class with a private IMEI number and display it using a public method.

class Mobile:
    def __init__(self, imei):
        self.__imei = imei  # Private variable

    def display_imei(self):
        print("IMEI Number:", self.__imei)


m = Mobile("123456789012345")
m.display_imei()
```

```python
#Write a program to create an ATM class with a private PIN and methods to verify and change the PIN securely.

class ATM:
    def __init__(self, pin):
        self.__pin = pin

    def verify_pin(self, pin):
        if self.__pin == pin:
            print("PIN Verified")
        else:
            print("Incorrect PIN")

    def change_pin(self, old_pin, new_pin):
        if self.__pin == old_pin:
            self.__pin = new_pin
            print("PIN Changed Successfully")
        else:
            print("Old PIN Incorrect")


atm = ATM(1234)

atm.verify_pin(1234)
atm.change_pin(1234, 5678)
atm.verify_pin(5678)
```

```python
#Write a program to create an abstract Payment class and implement payment processing for different payment methods.

from abc import ABC, abstractmethod

class Payment(ABC):

    @abstractmethod
    def process_payment(self):
        pass


class CreditCard(Payment):
    def process_payment(self):
        print("Payment processed using Credit Card")


class Banking(Payment):
    def process_payment(self):
        print("Payment processed using Banking App")


p1 = CreditCard()
p2 = Banking()

p1.process_payment()
p2.process_payment()
```

```python
#Write a program to create an abstract Animal class and print different sounds using child classes.

from abc import ABC, abstractmethod

class Animal(ABC):

    @abstractmethod
    def sound(self):
        pass


class Dog(Animal):
    def sound(self):
        print("Dog says: Bark")


class Cat(Animal):
    def sound(self):
        print("Cat says: Meow")


d = Dog()
c = Cat()

d.sound()
c.sound()
```

```python
#Write a program where SavingsAccount and CurrentAccount classes override the interest_rate() method of the Account class.

class Account:

    def interest_rate(self):
        print("General Interest Rate")


class SavingsAccount(Account):

    def interest_rate(self):
        print("Savings Account Interest Rate: 6%")


class CurrentAccount(Account):

    def interest_rate(self):
        print("Current Account Interest Rate: 2%")


s = SavingsAccount()
c = CurrentAccount()

s.interest_rate()
c.interest_rate()
```

```python
#Write a program where Student and Teacher classes override the role() method inherited from the Person class.

class Person:

    def role(self):
        print("I am a Person")


class Student(Person):

    def role(self):
        print("I am a Student")


class Teacher(Person):

    def role(self):
        print("I am a Teacher")


s = Student()
t = Teacher()

s.role()
t.role()
```
