```python
#Q1 Create a class Person with attributes name and age. Create a class Student that inherits from Person and adds a grade attribute. Display all details.

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display_person(self):
        print("Name :", self.name)
        print("Age  :", self.age)


class Student(Person):
    def __init__(self, name, age, grade):
        super().__init__(name, age)
        self.grade = grade

    def display_student(self):
        self.display_person()
        print("Grade:", self.grade)


s = Student("Relish", 18, "A")
s.display_student()
```

```python
#Q2 Create classes Teacher and Researcher, each with their own methods. Create a class Professor that inherits from both classes and demonstrates the functionality of both parents.

class Teacher:
    def teach(self):
        print("Teaching students.")


class Researcher:
    def research(self):
        print("Doing research.")


class Professor(Teacher, Researcher):
    def display(self):
        print("Professor's Work")


p = Professor()

p.display()
p.teach()
p.research()
```

```python
#Q3 Create a class hierarchy Animal → Mammal → Dog. Add one method in each class and show that a Dog object can access methods from all three classes.

class Animal:
    def eat(self):
        print("Animal eats food.")


class Mammal(Animal):
    def walk(self):
        print("Mammal walks.")


class Dog(Mammal):
    def bark(self):
        print("Dog barks.")


d = Dog()

d.eat()
d.walk()
d.bark()
```

```python
#Q4 Create a class Shape with a method display(). Create two child classes Circle and Rectangle that inherit from Shape and implement their own area calculation methods

class Shape:
    def display(self):
        print("Area Calculation")


class Circle(Shape):
    def area(self, radius):
        area = 3.14 * radius * radius
        print("Area of Circle =", area)


class Rectangle(Shape):
    def area(self, length, breadth):
        area = length * breadth
        print("Area of Rectangle =", area)


c = Circle()
c.display()
c.area(7)

print()

r = Rectangle()
r.display()
r.area(8, 5)
```
