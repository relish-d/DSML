```python
#Q1 student grade management system

class Student:
    # Class Attribute
    school = "ABC Secondary School"

    # Constructor
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    # Instance Method
    def check_grade(self):
        print("Student Name:", self.name)
        print("Marks:", self.marks)

        if self.marks >= 80:
            print("Grade: A")
        elif self.marks >= 60:
            print("Grade: B")
        elif self.marks >= 40:
            print("Grade: C")
        else:
            print("Result: Fail")

    # Class Method
    @classmethod
    def display_school(cls):
        print("School Name:", cls.school)

    # Static Method
    @staticmethod
    def welcome():
        print("Welcome to the Student Grade Management System!")


# Creating Object
student1 = Student("Relish", 85)

# Calling Static Method
Student.welcome()

# Calling Class Method
Student.display_school()

# Calling Instance Method
student1.check_grade()
```

```python
#Q2 Bank account managemnt system

class BankAccount:
    # Class Attribute
    bank_name = "RAD Bank"

    # Constructor
    def __init__(self, name, balance):
        self.name = name
        self.balance = balance

    # Instance Method - Deposit
    def deposit(self):
        amount = float(input("Enter deposit amount: "))
        self.balance += amount
        print("Current Balance:", self.balance)

    # Instance Method - Withdraw
    def withdraw(self):
        amount = float(input("Enter withdrawal amount: "))
        if amount <= self.balance:
            self.balance -= amount
            print("Current Balance:", self.balance)
        else:
            print("Insufficient Balance!")

    # Instance Method - Show Balance
    def show_balance(self):
        print("Current Balance:", self.balance)

    # Class Method
    @classmethod
    def display_bank(cls):
        print("Bank Name:", cls.bank_name)

    # Static Method
    @staticmethod
    def welcome():
        print("Welcome to RAD Bank")


# Main Program
BankAccount.welcome()
BankAccount.display_bank()

name = input("Enter account holder name: ")
balance = float(input("Enter initial balance: "))

account = BankAccount(name, balance)

account.deposit()
account.withdraw()
account.show_balance()
```
