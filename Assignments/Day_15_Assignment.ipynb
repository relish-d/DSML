```python
#Q1 Write a program that passes a function as an argument to another function and executes it.

def greet():
    print("Hello, Welcome!")

def execute(func):
    func()

execute(greet)
```

```python
#Q2 Write a program where a function returns another function that multiplies a number by 2

def multiplier():
    def multiply_by_two(num):
        return num * 2
    return multiply_by_two

double = multiplier()

print(double(10))
print(double(25))
```

```python
#Q3 Write a program that creates a decorator using nested functions and returns the wrapper function.

def decorator(func):
    def wrapper():
        print("Before function execution")
        func()
        print("After function execution")
    return wrapper

@decorator
def display():
    print("This is the original function.")

display()
```

```python
#Q4 Write a program using *args and **kwargs in a decorator so it can work with any function regardless of its parameters.

def decorator(func):
    def wrapper(*args, **kwargs):
        print("Function Started")
        result = func(*args, **kwargs)
        print("Function Finished")
        return result
    return wrapper

@decorator
def add(a, b):
    print("Sum =", a + b)

@decorator
def greet(name):
    print("Hello", name)

add(5, 8)
greet("Relish")
```

```python
#Q5 Write a custom iterator class EvenNumbers that generates even numbers from 2 up to a given limit.

class EvenNumbers:
    def __init__(self, limit):
        self.limit = limit
        self.current = 2

    def __iter__(self):
        return self

    def __next__(self):
        if self.current <= self.limit:
            value = self.current
            self.current += 2
            return value
        else:
            raise StopIteration

numbers = EvenNumbers(20)

for i in numbers:
    print(i)
```

```python
#Q6 Create an iterator class Countdown that starts from a given number and counts down to 1

class Countdown:
    def __init__(self, start):
        self.current = start

    def __iter__(self):
        return self

    def __next__(self):
        if self.current >= 1:
            value = self.current
            self.current -= 1
            return value
        else:
            raise StopIteration

count = Countdown(5)

for i in count:
    print(i)
```

```python
#Q7 Write a generator function square _generator(n) that yields squares of numbers from 1 to n.

def square_generator(n):
    for i in range(1, n + 1):
        yield i * i

for square in square_generator(10):
    print(square)
```
