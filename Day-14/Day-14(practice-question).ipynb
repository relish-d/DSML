```python
#Convert a list of names into lowercase using list comprehension

names = ["Ram", "Hari", "SITA", "Gita", "RAMESH"]

lowercase_names = [name.lower() for name in names]

print("Original List:", names)
print("Lowercase List:", lowercase_names)
```

```python
#Create a dictionary where each word is a key and its length is the value

words = ["apple", "banana", "cat", "elephant", "dog"]

word_length = {word: len(word) for word in words}

print(word_length)
```

```python
#Use a lambda function with filter() to extract all odd numbers from a list

numbers = [10, 15, 21, 30, 45, 52, 67]

odd_numbers = list(filter(lambda x: x % 2 != 0, numbers))

print("Odd Numbers:", odd_numbers)
```

```python
#Use a lambda function with map() to add 10 to every element in a list

numbers = [5, 10, 15, 20, 25]

new_numbers = list(map(lambda x: x + 10, numbers))

print("Original List:", numbers)
print("Updated List:", new_numbers)
```

```python
#Select student names with more than 4 characters and create a dictionary with their character count

students = ["Ram", "Hari", "Ramesh", "Sita", "Anushka", "Krishna", "Binod"]

selected_names = [name for name in students if len(name) > 4]

name_count = {name: len(name) for name in selected_names}

print("Selected Names:", selected_names)
print("Dictionary:", name_count)
```
