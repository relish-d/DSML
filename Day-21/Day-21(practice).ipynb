```python
#Create a base class Vehicle with a method move(). Create Car, Bike, and Plane classes that implement their own version of move(). Write a function start_journey(vehicle) that works with any vehicle object.

class Vehicle:
    def move(self):
        pass


class Car(Vehicle):
    def move(self):
        print("Car is moving on the road.")


class Bike(Vehicle):
    def move(self):
        print("Bike is moving on the road.")


class Plane(Vehicle):
    def move(self):
        print("Plane is flying in the sky.")


def start_journey(vehicle):
    vehicle.move()


car = Car()
bike = Bike()
plane = Plane()

start_journey(car)
start_journey(bike)
start_journey(plane)
```

```python
#Create a base class PaymentGateway with a method process_payment(amount). Create CreditCard, PayPal, and Esewa classes that implement their own payment processing logic. Write a function checkout(payment_method, amount) that works with any payment object.

class PaymentGateway:
    def process_payment(self, amount):
        pass


class CreditCard(PaymentGateway):
    def process_payment(self, amount):
        print(f"Rs.{amount} paid using Credit Card.")


class PayPal(PaymentGateway):
    def process_payment(self, amount):
        print(f"Rs.{amount} paid using PayPal.")


class Esewa(PaymentGateway):
    def process_payment(self, amount):
        print(f"Rs.{amount} paid using eSewa.")


def checkout(payment_method, amount):
    payment_method.process_payment(amount)


checkout(CreditCard(), 5000)
checkout(PayPal(), 3500)
checkout(Esewa(), 2000)
```

```python
#Create a base class Shape with a method area(). Create Rectangle, Circle, and Triangle classes that calculate their areas differently

class Shape:
    def area(self):
        pass


class Rectangle(Shape):
    def __init__(self, length, breadth):
        self.length = length
        self.breadth = breadth

    def area(self):
        return self.length * self.breadth


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius


class Triangle(Shape):
    def __init__(self, base, height):
        self.base = base
        self.height = height

    def area(self):
        return 0.5 * self.base * self.height


r = Rectangle(10, 5)
c = Circle(7)
t = Triangle(8, 4)

print("Rectangle Area:", r.area())
print("Circle Area:", c.area())
print("Triangle Area:", t.area())
```

```python
#Create a base class Employee with a method calculate_salary(). Create FullTimeEmployee and PartTimeEmployee classes that calculate salary using different formulas.

class Employee:
    def calculate_salary(self):
        pass


class FullTimeEmployee(Employee):
    def __init__(self, monthly_salary):
        self.monthly_salary = monthly_salary

    def calculate_salary(self):
        return self.monthly_salary


class PartTimeEmployee(Employee):
    def __init__(self, hours, rate):
        self.hours = hours
        self.rate = rate

    def calculate_salary(self):
        return self.hours * self.rate


f = FullTimeEmployee(50000)
p = PartTimeEmployee(120, 300)

print("Full Time Salary:", f.calculate_salary())
print("Part Time Salary:", p.calculate_salary())
```

```python
#Create classes PDFDocument, WordDocument, and ExcelDocument. Each class should have an open_file() method. Write a function open_document(doc) that opens any document type.

class PDFDocument:
    def open_file(self):
        print("Opening PDF document.")


class WordDocument:
    def open_file(self):
        print("Opening Word document.")


class ExcelDocument:
    def open_file(self):
        print("Opening Excel document.")


def open_document(doc):
    doc.open_file()


open_document(PDFDocument())
open_document(WordDocument())
open_document(ExcelDocument())
```

```python
#Create a base class Account with a method calculate_interest(amount). Create SavingsAccount, CurrentAccount, and FixedDepositAccount classes with different interest calculations

class Account:
    def calculate_interest(self, amount):
        pass


class SavingsAccount(Account):
    def calculate_interest(self, amount):
        return amount * 0.05


class CurrentAccount(Account):
    def calculate_interest(self, amount):
        return amount * 0.02


class FixedDepositAccount(Account):
    def calculate_interest(self, amount):
        return amount * 0.08


accounts = [
    SavingsAccount(),
    CurrentAccount(),
    FixedDepositAccount()
]

amount = 100000

for account in accounts:
    print(type(account).__name__, "Interest =", account.calculate_interest(amount))
```
