```python
#method  overriding
class PaymentGateway:
    def process_payment(self, amount):
        pass


class CreditCard(PaymentGateway):
    def process_payment(self, amount):
        return f"Processing ${amount} via Credit Card authentication."


class PayPal(PaymentGateway):
    def process_payment(self, amount):
        return f"Processing ${amount} via PayPal secure token."


# Unified interface function
def execute_checkout(payment_method, amount):
    print(payment_method.process_payment(amount))


# Usage
card = CreditCard()
wallet = PayPal()

execute_checkout(card, 250)
execute_checkout(wallet, 250)
```

```python
#operator overloading

class Point:
  def __init__(self,x,y):
    self.x = x
    self.y = y

  def __add__(self,other):
    x = self.x + other.x
    y = self.y + other.y
    return Point(x,y)

  def __sub__(self,other):
    x = self.x - other.x
    y = self.y - other.y
    return Point(x,y)
  def __mul__(self,other):
    x = self.x * other.x
    y = self.y * other.y
    return Point(x,y)

  def __str__(self):
    return f"({self.x},{self.y})"
```

```python
#create point object

p1 = Point(1,2)
p2 = Point(3,4)
```

```python
p3 = p1 + p2
print(p3)
```

```python
p4 = p1 - p2
print(p4)
```

```python
#operator overloading

class Point:
  def __init__(self,x):
    self.x = x

  def __mul__(self,other):
    x = self.x * other.x
    return Point(x)

  def __str__(self):
    return f"({self.x})"

p5 = Point(6)
p6 = Point(7)
p7 = p5 * p6
print(p7)
```
